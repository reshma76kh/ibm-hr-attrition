```python


def visualization(df, column):

    print("-" * 40)
    print("Visualization")
    print("-" * 40)

    # Histogram
    print("\nHistogram")

    plt.figure(figsize=(6,4))
    plt.hist(df[column], bins=15)
    plt.title(f"Histogram of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Employees")
    plt.grid(True)
    plt.show()

    # KDE Plot
    print("\nKDE Plot")

    plt.figure(figsize=(6,4))
    df[column].plot(kind='kde')
    plt.title(f"KDE Plot of {column}")
    plt.xlabel(column)
    plt.grid(True)
    plt.show()

    # Skewness
    print("\nSkewness")
    print(df[column].skew())

    # Box Plot
    print("\nBox Plot")

    plt.figure(figsize=(4,6))
    df[column].plot(kind='box')
    plt.title(f"Box Plot of {column}")
    plt.ylabel(column)
    plt.grid(True)
    plt.show()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    BaggingClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)
from sklearn.naive_bayes import MultinomialNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier

```

```python
svc = SVC(kernel='sigmoid', gamma=1.0)
knc = KNeighborsClassifier()
dtc = DecisionTreeClassifier(max_depth=5, random_state=42)
lrc = LogisticRegression(solver='liblinear', penalty='l1', random_state=42)
rfc = RandomForestClassifier(n_estimators=50, random_state=42)
abc = AdaBoostClassifier(n_estimators=50, random_state=42)
bc = BaggingClassifier(n_estimators=50, random_state=42)
etc = ExtraTreesClassifier(n_estimators=50, random_state=42)
gbdt = GradientBoostingClassifier(n_estimators=50, random_state=42)
xgb = XGBClassifier(n_estimators=50, random_state=42)
```

```python
Model = {
    'SVC' : svc,
    'KN' : knc,
    'DT': dtc,
    'LR': lrc,
    'RF': rfc,
    'AdaBoost': abc,
    'BgC': bc,
    'ETC': etc,
    'GBDT':gbdt,
    'xgb':xgb
}
```

```python
def evaluate_models(Model,x_train,y_train,x_test,y_test):
  result=[]
  for name ,model in Model.items():
    model.fit(x_train,y_train)
    y_pred=model.predict(x_test)

    result.append({
            "Model": name,
            "Accuracy": accuracy_score(y_test, y_pred),
            "Precision": precision_score(y_test, y_pred),
            "Recall": recall_score(y_test, y_pred),
            "F1 Score": f1_score(y_test, y_pred)
        })
  return pd.DataFrame(result)
```

```python
result = evaluate_models(
    Model,
    x_train_smote,
    y_train_smote,
    x_test,
    y_test
)

result.sort_values(by="F1 Score", ascending=False)
```

Among all evaluated models, AdaBoost achieved the highest F1-score (0.50), indicating the best balance between precision and recall. While XGBoost achieved the highest overall accuracy (86.7%), its recall for employees who leave was relatively low. Since the business objective is to identify employees at risk of attrition, AdaBoost is the most suitable baseline model
